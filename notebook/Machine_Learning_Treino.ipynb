## **Diagnóstico médico: Problemas Ortopédicos na Coluna Vertebral**

Esse dataset contém dados sobre problemas ortopédicos na coluna vertebral, diagnosticados no Centre Médico-Chirurgical de Réadaptation des Massues, em Lyon, France. Contém 6 atributos biomecânicos para 310 entradas anonimizadas, sendo 100 de pacientes considerados sem problemas (Normal - NO), 60 de pacientes com Hérnia de Disco (Disk Hernia - DH) e 150 de pacientes com Espondilolistese (Spondylolisthesis - SL).

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["diagnostic"] = le.fit_transform(df["diagnostic"])
```

```python
df["diagnostic"].value_counts()
```

Crie as variáveis X e y.

X deve ser um dataframe que contenha os dados dos seis atributos referentes aos biomarcadores (V1 a V6).

y deve ser um array com os dados da coluna "diagnostic".

```python
#X = atributos
X = df[["V1", "V2", "V3", "V4", "V5", "V6"]]
```

```python
#y = alvo
y = df["diagnostic"]
```

```python
#primeiros registros de X
X.head()
```

```python
#X como numpy
X.values
```

```python
#y como numpy
y.values
```

Abaixo, separe 70% dos dados para treinamento e 30% para teste.

Utilize os nomes X_train, X_test, y_train e y_test


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
#criando modelo
modelo1 = DecisionTreeClassifier()
```

```python
#treinando
modelo1.fit(X_train, y_train)
```

```python
#previsões
y_predict_1 = modelo1.predict(X_test)
y_predict_1
```

```python
from sklearn.neighbors import KNeighborsClassifier
#criando modelo
modelo2 = KNeighborsClassifier()
```

```python
#treinar
modelo2.fit(X_train, y_train)
```

```python
#previsões
y_predict_2 = modelo2.predict(X_test)
```

```python
y_predict_2
```

```python
from sklearn.metrics import accuracy_score
#acurácia modelo 1
acc1 = accuracy_score(y_test, y_predict_1)
print("Acurácia Modelo 1:", acc1)
```

```python
from sklearn.metrics import accuracy_score
#acurácia modelo 2
acc2 = accuracy_score(y_test, y_predict_2)
print("Acurácia Modelo 2:", acc2)
```

```python
#escolhendo modelo
if acc1 > acc2:
    modelo = modelo1
else:
    modelo = modelo2
```

```python
import os
import pickle
#criando a pasta
os.makedirs("model", exist_ok=True)
#salvando modelo
with open("model/modelo.pkl", "wb") as f:
    pickle.dump(modelo, f)
```

## Você pode construir um modelo de aprendizado de máquina para prever com precisão se os pacientes no conjunto de dados têm diabetes ou não?

Esse dataset contém dados sobre incidência de diabetes em mulheres do Povo Pima (índios nativos norte americanos originários do atual Estado do Arizona). Contém 8 atributos biomédicos para 768 entradas anonimizadas, sendo 500 testados negativo para diabetes e 268 testados positivo para diabetes.

```python
#Carregando o dataser
import pandas as pd
df_diabetes = pd.read_csv("diabetes.csv")
df_diabetes.head()
```

```python
#Verificar tamanho
df_diabetes.shape
```

```python
#Carregar informações
df_diabetes.info()
```

```python
#Carregar estatísticas
df_diabetes.describe()
```

```python
# X = todas as colunas
X = df_diabetes.drop("Outcome", axis=1)
# y = coluna alvo
y = df_diabetes["Outcome"]
```

```python
#Partindo pra visualização:
```

```python
X.head()
```

```python
X.values
```

```python
y.values
```

```python
#Separando treino
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
#Modelo 1: Árvore de decisão
from sklearn.tree import DecisionTreeClassifier
modelo1 = DecisionTreeClassifier()
modelo1.fit(X_train, y_train)
y_predict_1 = modelo1.predict(X_test)
```

```python
#Modelo 2: KNN
from sklearn.neighbors import KNeighborsClassifier
modelo2 = KNeighborsClassifier()
modelo2.fit(X_train, y_train)
y_predict_2 = modelo2.predict(X_test)
```

```python
#Visualizando previsões:
```

```python
y_predict_1
```

```python
y_predict_2
```

```python
#Acurácia
from sklearn.metrics import accuracy_score
acc1 = accuracy_score(y_test, y_predict_1)
print("Acurácia Modelo 1:", acc1)
acc2 = accuracy_score(y_test, y_predict_2)
print("Acurácia Modelo 2:", acc2)
```

```python
#escolhendo o melhor modelo
if acc1 > acc2:
    modelo = modelo1
else:
    modelo = modelo2
```

```python
import os
import pickle
os.makedirs("model", exist_ok=True)
with open("model/modelo_diabetes.pkl", "wb") as f:
    pickle.dump(modelo, f)
```
